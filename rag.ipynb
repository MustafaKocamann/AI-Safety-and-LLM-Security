## Fake KB

```python
from fake_docs import DOCUMENTS

console = Console()

table = Table(
    title="📚 Novacure Bilgi Tabanı (Knowledge Base)",
    show_header=True,
    header_style="bold magenta",
    border_style="cyan",
    expand=True
)

table.add_column("ID", style="bold yellow", width=12, justify="center")
table.add_column("Başlık (Title)", style="bold green", width=22)
table.add_column("İçerik (Text)", style="white")

for doc in DOCUMENTS:
    table.add_row(doc["id"], doc["title"], doc["text"])

console.print(Panel(table, border_style="blue", padding=(1, 2)))
```

### LangChain works with Document objects

```python
from langchain_core.documents import Document

documents = [
    Document(page_content=d["text"], metadata={"id": d["id"], "title": d["title"]})
    for d in DOCUMENTS
]
documents[0]
```

## Create embeddings and the vector store

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore

# Hugging Face Embedding Modeli
embeddings = HuggingFaceEmbeddings(
    model_name=EMBEDDING_MODEL,
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)
```

```python

from langchain_core.vectorstores import InMemoryVectorStore

# Query embedding oluştur
vector = embeddings.embed_query(
    "How many leave days do I get?"
)

print(
    "One embedding has",
    len(vector),
    "numbers. First five:",
    [round(x, 4) for x in vector[:5]]
)

# In-memory vector database oluştur
vector_store = InMemoryVectorStore(embeddings)

# Dokümanları vector store'a ekle
vector_store.add_documents(documents)

print("Documents stored:", len(documents))
```
