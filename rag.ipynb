## Fake KB

```python
from fake_docs import DOCUMENTS

console = Console()

table = Table(
    title="📚 Novacure Bilgi Tabanı (Knowledge Base)",
    show_header=True,
    header_style="bold magenta",
    border_style="cyan",
    expand=True
)

table.add_column("ID", style="bold yellow", width=12, justify="center")
table.add_column("Başlık (Title)", style="bold green", width=22)
table.add_column("İçerik (Text)", style="white")

for doc in DOCUMENTS:
    table.add_row(doc["id"], doc["title"], doc["text"])

console.print(Panel(table, border_style="blue", padding=(1, 2)))
```

### LangChain works with Document objects

```python
from langchain_core.documents import Document

documents = [
    Document(page_content=d["text"], metadata={"id": d["id"], "title": d["title"]})
    for d in DOCUMENTS
]
documents[0]
```

## Create embeddings and the vector store

```python
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_google_genai import GoogleGenerativeAIEmbeddings

console = Console()

# 1. Google Embedding Modelini Tanımlama
EMBEDDING_MODEL = "embedding-001"
embeddings = GoogleGenerativeAIEmbeddings(model=f"models/{EMBEDDING_MODEL}")

# 2. Örnek Sorgu Embedding'i Alma
query_text = "How many leave days do I get?"
vector = embeddings.embed_query(query_text)
first_five = [round(x, 4) for x in vector[:5]]

# 3. In-Memory Vector Store'a Belgeleri Yükleme
vector_store = InMemoryVectorStore(embeddings)

from langchain_core.documents import Document
langchain_docs = [
    Document(page_content=doc["text"], metadata={"id": doc["id"], "title": doc["title"]})
    for doc in DOCUMENTS
]

vector_store.add_documents(langchain_docs)

# --- 4.PRETTY PRINT---
table = Table(title="🔍 Vektör Veritabanı ve Embedding Durumu", show_header=True, header_style="bold cyan", border_style="blue")
table.add_column("İşlem / Metrik", style="bold yellow", width=25)
table.add_column("Detay / Değer", style="white")

table.add_row("Kullanılan Embedding Model", f"Google GenAI ({EMBEDDING_MODEL})")
table.add_row("Test Sorgusu", f'"{query_text}"')
table.add_row("Vektör Boyutu (Dimension)", str(len(vector)))
table.add_row("İlk 5 Vektör Değeri", str(first_five))
table.add_row("Depolanan Belge Sayısı", f"✅ {len(langchain_docs)} adet belge başarıyla eklendi")

console.print(Panel(table, title="[bold green]Vector Store Başarıyla Oluşturuldu[/bold green]", border_style="magenta"))
```
