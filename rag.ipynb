```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# Çevresel değişkenleri yükle
load_dotenv()

# 1. API Anahtarlarının Doğrulanması
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

assert GROQ_API_KEY, "❌ GROQ_API_KEY eksik. Lütfen .env dosyasına ekleyin."
assert GOOGLE_API_KEY, "❌ GOOGLE_API_KEY eksik. Lütfen .env dosyasına ekleyin."

# (LangChain, GOOGLE_API_KEY ortam değişkenini otomatik olarak okur)
os.environ["GOOGLE_API_KEY"] = GOOGLE_API_KEY

# 2. Model ve Servis Tanımlamaları
CHAT_MODEL = os.getenv("CHAT_MODEL", "gpt-oss-120b")
EMBEDDING_MODEL = os.getenv("EMBEDDING_MODEL", "embedding-001")

# 3. İstemci Yapılandırmaları
# Groq (OpenAI Uyumlu İstemci)
groq_client = OpenAI(
    api_key=GROQ_API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

# LangChain Google Embedding Modeli
embeddings = GoogleGenerativeAIEmbeddings(
    model=f"models/{EMBEDDING_MODEL}"
)

# 4. Profesyonel Pretty Print Çıktısı
try:
    from rich.console import Console
    from rich.panel import Panel
    from rich.table import Table

    console = Console()
    
    table = Table(title="🚀 Hibrit Yapay Zeka Altyapısı (Groq + LangChain Google)", show_header=True, header_style="bold cyan")
    table.add_column("Bileşen", style="dim", width=16)
    table.add_column("Sağlayıcı / Çerçeve", style="green", width=22)
    table.add_column("Model Adı", style="magenta")
    table.add_column("Durum", style="bold green", justify="center")

    table.add_row("Chat (LLM)", "Groq (OpenAI Uyumlu)", CHAT_MODEL, "🟢 Aktif")
    table.add_row("Embedding", "LangChain Google GenAI", EMBEDDING_MODEL, "🟢 Aktif")

    console.print(Panel(table, title="[bold yellow]Sistem Başarıyla Başlatıldı[/bold yellow]", border_style="blue"))

except ImportError:
    # Rich kütüphanesi yoksa alternatif düz metin çıktısı
    print("=" * 65)
    print(" 🚀 HİBRİT YAPAY ZEKA ALTYAPISI BAŞLATILDI ")
    print("=" * 65)
    print(f" • Chat LLM (Groq)             : {CHAT_MODEL}")
    print(f" • Embedding (LangChain/Google): {EMBEDDING_MODEL}")
    print(f" • Durum                       : Tüm servisler hazır ✅")
    print("=" * 65)
```
